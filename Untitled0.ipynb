## 1. Persiapan: koneksi dan buat tabel

Saya pakai `sqlite3` bawaan Python, sesuai Modul 9.2. `PRAGMA foreign_keys = ON` saya nyalakan supaya foreign key benar-benar dijaga. Kalau notebook di-run ulang, tabel lama saya hapus dulu supaya tidak duplikat.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect('perpustakaan.db')
cur = conn.cursor()
cur.execute('PRAGMA foreign_keys = ON;')

cur.execute('DROP TABLE IF EXISTS peminjaman')
cur.execute('DROP TABLE IF EXISTS buku')
cur.execute('DROP TABLE IF EXISTS anggota')

cur.execute('''CREATE TABLE buku (
    id INTEGER PRIMARY KEY,
    judul TEXT NOT NULL,
    penulis TEXT NOT NULL,
    tahun INTEGER
)''')
cur.execute('''CREATE TABLE anggota (
    id INTEGER PRIMARY KEY,
    nama TEXT NOT NULL,
    email TEXT NOT NULL
)''')
cur.execute('''CREATE TABLE peminjaman (
    id INTEGER PRIMARY KEY,
    id_anggota INTEGER NOT NULL,
    id_buku INTEGER NOT NULL,
    tgl_pinjam TEXT NOT NULL,
    tgl_kembali TEXT,
    FOREIGN KEY (id_anggota) REFERENCES anggota(id),
    FOREIGN KEY (id_buku) REFERENCES buku(id)
)''')
conn.commit()
print('Tabel buku, anggota, dan peminjaman berhasil dibuat.')
```

## 1a. Tambah 5 buku dan 3 anggota

```python
buku = [
    (1, 'Laskar Pelangi', 'Andrea Hirata', 2005),
    (2, 'Bumi Manusia', 'Pramoedya Ananta Toer', 1980),
    (3, 'Dilan: Dia adalah Dilanku Tahun 1990', 'Pidi Baiq', 2014),
    (4, 'Filosofi Kopi', 'Dee Lestari', 2006),
    (5, 'Ayat-Ayat Cinta', 'Habiburrahman El Shirazy', 2004),
]
anggota = [
    (1, 'Alya Zhafira', 'alya.zhafira@example.com'),
    (2, 'Alya Dhiya Oktaviani', 'alya.dhiya@example.com'),
    (3, 'Salma Luqyana Gunawan', 'salma.luqyana@example.com'),
]
cur.executemany('INSERT INTO buku VALUES (?, ?, ?, ?)', buku)
cur.executemany('INSERT INTO anggota VALUES (?, ?, ?)', anggota)
conn.commit()
print('5 buku dan 3 anggota berhasil ditambahkan.')
print(pd.read_sql_query('SELECT * FROM buku', conn))
print(pd.read_sql_query('SELECT * FROM anggota', conn))
```

## 1b. 4 transaksi peminjaman

Satu peminjaman sengaja saya biarkan belum dikembalikan (`tgl_kembali` NULL): Alya Dhiya Oktaviani meminjam *Bumi Manusia* tanggal 3 November 2026.

```python
peminjaman = [
    (1, 1, 1, '2026-11-02', '2026-11-09'),  # Alya Zhafira - Laskar Pelangi, 7 hari
    (2, 1, 3, '2026-11-05', '2026-11-12'),  # Alya Zhafira - Dilan, 7 hari
    (3, 2, 2, '2026-11-03', None),           # Alya Dhiya Oktaviani - Bumi Manusia, belum kembali
    (4, 3, 4, '2026-11-06', '2026-11-10'),  # Salma Luqyana Gunawan - Filosofi Kopi, 4 hari
]
cur.executemany('INSERT INTO peminjaman VALUES (?, ?, ?, ?, ?)', peminjaman)
conn.commit()
print(pd.read_sql_query('SELECT * FROM peminjaman', conn))
```

## 1c. Daftar peminjaman lengkap (JOIN)

Saya pakai `INNER JOIN` dua kali: peminjaman ke anggota untuk nama, dan ke buku untuk judul. Hasilnya diurutkan berdasarkan tanggal pinjam.

```python
query_daftar = '''
SELECT p.id AS id_peminjaman, a.nama, b.judul, p.tgl_pinjam, p.tgl_kembali
FROM peminjaman p
JOIN anggota a ON p.id_anggota = a.id
JOIN buku b ON p.id_buku = b.id
ORDER BY p.tgl_pinjam
'''
df_daftar = pd.read_sql_query(query_daftar, conn)
df_daftar
```

## 1d. Total buku yang dipinjam tiap anggota (GROUP BY + COUNT)

Alya Zhafira meminjam paling banyak, 2 buku. Alya Dhiya Oktaviani dan Salma Luqyana Gunawan masing-masing 1 buku.

```python
query_total = '''
SELECT a.nama, COUNT(p.id) AS total_buku
FROM anggota a
LEFT JOIN peminjaman p ON p.id_anggota = a.id
GROUP BY a.id, a.nama
ORDER BY total_buku DESC
'''
df_total = pd.read_sql_query(query_total, conn)
df_total
```

## 2a. DataFrame pandas dari hasil query SQL

Saya ambil lagi hasil 1c, tapi kali ini sebagai DataFrame pandas supaya bisa diolah lebih lanjut. Ini DataFrame yang sama dengan tabel di 1c.

```python
df = pd.read_sql_query(query_daftar, conn)
df
```

## 2b. Eksplorasi data: durasi peminjaman

Durasi saya hitung sebagai `tgl_kembali - tgl_pinjam` dalam hari. Peminjaman yang belum kembali (Alya Dhiya Oktaviani) durasinya jadi NaN, jadi saya tidak ikutkan dalam rata-rata — kalau dipaksa diisi, rata-ratanya jadi menyesatkan.

Hasilnya: 3 peminjaman yang sudah kembali berdurasi 7, 7, dan 4 hari. Rata-ratanya 6 hari, mediannya 7 hari. Polanya: Alya Zhafira yang paling aktif (2 peminjaman), dan semua peminjaman yang sudah kembali selesai dalam seminggu atau kurang.

```python
df['tgl_pinjam'] = pd.to_datetime(df['tgl_pinjam'])
df['tgl_kembali'] = pd.to_datetime(df['tgl_kembali'])
df['durasi_hari'] = (df['tgl_kembali'] - df['tgl_pinjam']).dt.days
print(df[['nama', 'judul', 'tgl_pinjam', 'tgl_kembali', 'durasi_hari']])
print()
print('Statistik durasi (hanya yang sudah kembali):')
print(df['durasi_hari'].describe())
print()
print('Jumlah peminjaman per anggota:')
print(df['nama'].value_counts())
```

## 2c. Visualisasi (matplotlib)

Dua gambar: histogram durasi peminjaman, dan diagram batang jumlah buku yang dipinjam per anggota. Histogramnya hanya dari 3 data yang sudah kembali, jadi bentuknya sederhana — ini saya sampaikan jujur, bukan saya tambah-tambahi datanya.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

df['durasi_hari'].hist(ax=ax[0], bins=3)
ax[0].set_title('Histogram Durasi Peminjaman (hari)')
ax[0].set_xlabel('Durasi (hari)')
ax[0].set_ylabel('Jumlah peminjaman')

df_total.plot.bar(x='nama', y='total_buku', ax=ax[1], legend=False)
ax[1].set_title('Jumlah Buku yang Dipinjam per Anggota')
ax[1].set_xlabel('Anggota')
ax[1].set_ylabel('Total buku')
plt.tight_layout()
plt.show()
```
